```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        item = {"name": name, "price": price}
        self.items.append(item)
        print(f"Đã thêm '{name}' vào giỏ hàng.")
    def remove_item(self, name):
        found = False
        for item in self.items:
            if item["name"].lower() == name.lower():
                self.items.remove(item)
                print(f"Đã xóa '{name}' khỏi giỏ hàng.")
                found = True
                break
        if not found:
            print(f"Không tìm thấy sản phẩm '{name}' để xóa.")

    def is_empty(self):
        # 4. Kiểm tra xem giỏ hàng có rỗng không
        return len(self.items) == 0

    def calculate_total(self):
        # 5. Tính tổng tiền giỏ hàng
        total = sum(item["price"] for item in self.items)
        return total

    def display_cart(self):
        # 6. Hiển thị toàn bộ sản phẩm trong giỏ
        if self.is_empty():
            print("Giỏ hàng đang trống.")
        else:
            print("--- Các sản phẩm trong giỏ ---")
            for i, item in enumerate(self.items, 1):
                print(f"{i}. {item['name']}: {item['price']:,} VNĐ")

    def clear_cart(self):
        # 7. Xóa toàn bộ giỏ hàng
        self.items = []
        print("Đã xóa toàn bộ giỏ hàng.")

# --- Phần Menu Chương Trình ---

def main():
    cart = ShoppingCart()

    while True:
        print("\n--- MENU GIỎ HÀNG ---")
        print("2. Thêm sản phẩm")
        print("3. Xóa sản phẩm theo tên")
        print("4. Kiểm tra giỏ hàng rỗng")
        print("5. Tính tổng tiền")
        print("6. Hiển thị giỏ hàng")
        print("7. Xóa toàn bộ giỏ hàng")

        choice = input("Nhập lựa chọn của bạn (2-7): ")

        if choice == '2':
            name = input("Nhập tên sản phẩm: ")
            price = float(input("Nhập giá sản phẩm: "))
            cart.add_item(name, price)
        elif choice == '3':
            name = input("Nhập tên sản phẩm cần xóa: ")
            cart.remove_item(name)
        elif choice == '4':
            if cart.is_empty():
                print("Giỏ hàng đang rỗng.")
            else:
                print("Giỏ hàng đang có sản phẩm.")
        elif choice == '5':
            print(f"Tổng tiền hóa đơn: {cart.calculate_total():,} VNĐ")
        elif choice == '6':
            cart.display_cart()
        elif choice == '7':
            cart.clear_cart()
        else:
            print("Lựa chọn không hợp lệ.")

        # Hỏi thoát chương trình
        quit_confirm = input("Có muốn thoát khỏi chương trình không? (nhập 'yes' để thoát): ").lower()
        if quit_confirm == 'yes':
            print("Cảm ơn bạn đã sử dụng dịch vụ!")
            break
```

```python
from math import *
class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Mẫu số phải khác 0")
        # Tìm ước chung lớn nhất để rút gọn
        common = gcd(num, den)
        self.num = num // common
        self.den = den // common
        # Quy ước mẫu số luôn dương
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den
    def add(self, other):
        # a/b + c/d = (ad + bc) / bd
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def sub(self, other):
        # a/b - c/d = (ad - bc) / bd
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def mul(self, other):
        # a/b * c/d = (ac) / (bd)
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def div(self, other):
        # a/b / c/d = (ad) / (bc)
        if other.num == 0:
            raise ZeroDivisionError("Không thể chia cho phân số có tử bằng 0")
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)
    def __str__(self):
        # Nếu mẫu bằng 1 thì chỉ in tử số, hoặc in dạng a/b
        if self.den == 1:
            return str(self.num)
        return f"{self.num}/{self.den}"
# --- Xử lý Input theo định dạng: a b op c d ---
def solve():
        line = input().split()
        a, b = int(line[0]), int(line[1])
        op = line[2]
        c, d = int(line[3]), int(line[4])
        f1 = Fraction(a, b)
        f2 = Fraction(c, d)
        if op == '+':
            result = f1.add(f2)
        elif op == '-':
            result = f1.sub(f2)
        elif op == '*':
            result = f1.mul(f2)
        elif op == '/':
            result = f1.div(f2)
        else:
            return "Toán tử không hợp lệ"

        print(result)
solve()
#Khai báo đối tượng other để gọi đối tượng khác.... f1.add(f2)
```

```python
class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        # Khởi tạo phần thực và phần ảo
        self.realPart = float(realPart)
        self.imaginaryPart = float(imaginaryPart)

    def multiply(self, other):
        # Công thức nhân số phức: (a+bi)(c+di) = (ac-bd) + (ad+bc)i
        new_real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        new_imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(new_real, new_imag)

    def __str__(self):
        # In ra phần thực và ảo cách nhau bởi dấu cách
        # Sử dụng int() nếu kết quả là số nguyên để khớp định dạng đầu ra mẫu
        r = int(self.realPart) if self.realPart.is_integer() else self.realPart
        i = int(self.imaginaryPart) if self.imaginaryPart.is_integer() else self.imaginaryPart
        return f"{r} {i}"

# Xử lý nhập liệu: mỗi dòng là một số phức
try:
    while True:
        line1 = input().split()
        if not line1: break
        line2 = input().split()

        c1 = ComplexNumber(line1[0], line1[1])
        c2 = ComplexNumber(line2[0], line2[1])

        print(c1.multiply(c2))
except EOFError:
    pass
```

```python
import math

class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.a = int(edge1)
        self.b = int(edge2)
        self.c = int(edge3)

    def is_valid(self):
        # Điều kiện tồn tại tam giác: tổng 2 cạnh bất kỳ lớn hơn cạnh còn lại
        return (self.a + self.b > self.c and
                self.a + self.c > self.b and
                self.b + self.c > self.a)

    def getArea(self):
        if not self.is_valid():
            return -1.0

        # Công thức Heron
        s = (self.a + self.b + self.c) / 2
        area = math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))
        return area

# Đọc input và in kết quả
import sys
for line in sys.stdin:
    try:
        edges = list(map(int, line.split()))
        if len(edges) != 3:
            print("invalid")
            continue

        tri = Triangle(edges[0], edges[1], edges[2])
        area = tri.getArea()

        if area == -1.0:
            print("invalid")
        else:
            # Làm tròn đến 2 chữ số thập phân
            if area.is_integer():
                print(int(area))
            else:
                print(f"{area:.2f}")
    except EOFError:
        break
```
