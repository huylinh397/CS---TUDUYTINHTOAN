```python

  try:
    n = int(input())
    def check(n):
      seen = set() # Dùng set để kiểm tra vòng lặp vô hạn

      while n != 1:
              # Tính tổng bình phương các chữ số
        current_sum = 0
        for digit in str(n):
          current_sum += int(digit) ** 2

              # Kiểm tra xem kết quả này đã từng xuất hiện chưa
        if current_sum in seen:
          return print("NO") # Rơi vào vòng lặp vô hạn
        seen.add(current_sum)
        n = current_sum
      return print('YES')
    check(n)
  except:
    print('ValueError')


```

```python
import re

def check_regex():
    try:
        t = int(input())  # Đọc số lượng testcase
        for _ in range(t):
            s = input()   # Đọc chuỗi S
            try:
                re.compile(s)  # Thử biên dịch chuỗi S thành Regex
                print("True")
            except re.error:   # Nếu có lỗi cú pháp Regex
                print("False")
    except EOFError:
        pass

check_regex()
```
